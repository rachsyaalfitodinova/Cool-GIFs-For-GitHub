Tantangan logika tukar isi gelas (variabel swapping)

```python

# Keadaan Awal
gelas_A = "Kopi"
gelas_B = "Teh"

# 3 baris logika penukaran
gelas_kosong = gelas_A
gelas_A = gelas_B
gelas_B = gelas_kosong

# Cek Hasil Akhir
print(f"Isi Gelas A sekarang: {gelas_A}")
print(f"Isi Gelas B sekarang: {gelas_B}")
```
