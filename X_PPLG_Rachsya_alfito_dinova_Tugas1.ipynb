```python

%%writefile praktik_variabel.cpp
#include <iostream>
#include <string>
using namespace std;

int main() {

    // Deklarasi Variabel dan Tipe Data
    string namaBarang = "iPhone 18 Pro";
    int jumlah = 2;
    float harga = 52000000.00;
    char kategori = 'A';
    bool tersedia = true;

    // Menampilkan Output ke Layar
    cout << "=== DATA INVENTARIS ===" << endl;
    cout << "Nama Barang : " << namaBarang << endl;
    cout << "Kategori    : " << kategori << endl;
    cout << "Jumlah      : " << jumlah << endl;
    cout << "Harga       : Rp " << harga << endl;
    cout << "Tersedia    : " << tersedia << " (1 = Ya, 0 = Tidak)" << endl;
    cout << "=======================" << endl;

    return 0;
}
```
