Membuat variabel penjumlahan dan pengurangan

```python

%%writefile praktik_variabel.cpp
#include <iostream>
using namespace std;

int main() {
    int bilangan1, bilangan2;

    cout << "Masukkan bilangan pertama: ";
    cin >> bilangan1;

    cout << "Masukkan bilangan kedua: ";
    cin >> bilangan2;

    cout << "Hasil penjumlahan: " << bilangan1 + bilangan2 << endl;
    cout << "Hasil pengurangan: " << bilangan1 - bilangan2 << endl;

    return 0;
}
```

```python

!g++ praktik_variabel.cpp -o praktik_variabel
!./praktik_variabel
```
