Tugas membuat program menghitung keliling persegi panjang

```python

%%writefile praktik_keliling.cpp
#include <iostream>
using namespace std;

int main() {
    int p, l, keliling;

    cout << "Masukkan panjang: ";
    cin >> p;

    cout << "Masukkan lebar: ";
    cin >> l;

    keliling = 2 * (p + l);

    cout << "Hasil keliling persegi panjang: " << keliling;

    return 0;
}
```

```python

!g++ praktik_keliling.cpp -o praktik_keliling
!./praktik_keliling
```
