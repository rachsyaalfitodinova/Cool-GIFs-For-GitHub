Tantangan membuat sistem pembelian tiket wahana roller coaster

```python

print("=== SISTEM TIKET ROLLER COASTER ===")

umur = int(input("Masukkan Umur Pengunjung: "))

if umur >= 12:
    tinggi = int(input("Masukkan Tinggi Badan (cm): "))

    if tinggi >= 145:
        print("Boleh Naik Wahana!")
    else:
        print("Maaf, umur cukup tapi tinggi badan belum memenuhi syarat.")
else:
    print("Maaf, belum cukup umur!")
```
